```python
# 문제 52. groupby와 정렬 함께 쓰기
# /data/happy.csv에서 년도별 점수 평균을 구하고, 평균 점수가 높은 순으로 정렬한 뒤 첫 번째 년도를 출력하라.

import pandas as pd

data_path = "/data/happy.csv"
df = pd.read_csv(data_path)

#사용자입력

res = df.groupby("년도")["점수"].mean()
print(res.sort_values(ascending=False).index[0])

#풀이코드
result = df.groupby("년도")["점수"].mean().sort_values(ascending=False).reset_index()
print(result.loc[0, "년도"])
```

```python
# 문제 53. 필터링 후 정렬하기
# /data/worldcup.csv에서 Country가 Brazil인 선수만 고른 뒤, Goals는 내림차순, Player는 오름차순으로 정렬해서 앞 3개 Player를 리스트로 출력하라.

import pandas as pd

data_path = "/data/worldcup.csv"
df = pd.read_csv(data_path)

#사용자입력

res = df[(df["Country"] == "Brazil")]
fin = res.sort_values(by= ["Goals", "Player"], ascending=[False, True]).head(3)
print(fin["Player"].tolist())

#풀이코드

result = df[df["Country"] == "Brazil"].sort_values(["Goals", "Player"], ascending=[False, True])
print(list(result["Player"].head(3)))
```
